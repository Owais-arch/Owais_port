```python
def intrest_cal(prin , rate , time): # prin , rate , time  nothing local and global
    """
    this pgm calculates simple intrest
    where time is in  year
    rate is in decimal   example 7 % is given as .07
    prin is any postive number 
    """
    print("principal" , prin) 
    global amount, intrest  #  10000 , 1000
    intrest = prin * rate * time
    amount = prin + intrest
    print("intrest is ", intrest)
    print("amount is " , amount)   
```

```python
def intrest_cals(prin , rate , time):
    """
    this pgm calculates simple intrest
    where time is in  year
    rate is in decimal   example 7 % is given as .07
    prin is any postive number 
    """
    print("principal" , prin)  #  2000000, 40000
    
    intrest = prin * rate * time
    amount = prin + intrest
    print("intrest is ", intrest)
    print("amount is " , amount) 
```

```python
#  def pgm to find the factorial of a number 
def fact_number( number ):
    """
    this pgm will find factorial 
    example factorial of 5 is 5*4*3*2*1
    """
    if number < 0 :
        return(" enter a postive integer")
    fact = 1    
    for i in range(1 , number+1  ):
        fact = fact * i
    return fact    
        
```

```python
def fact_numbers( number ):
    """
    this pgm will find factorial 
    example factorial of 5 is 5*4*3*2*1
    """
    if number < 0 :
        return(" enter a postive integer")
    fact = 1    
    for i in range(1 , number+1  ):
        fact = fact * i
    return fact  
```

```python
fact_number(5)
```

```python
# def pgm to find if number is prime or not

def prime_number( number):
    if number < 1 :
        return ("enter a integer greater than 1")
    if type(number ) == float:
        return ("enter a integer greater than 1")
    for i in range(2 , number):
        if number % i == 0:
            return(number  , "not prime")
    else   :
        return(number  , "is  prime")   
```

```python
prime_number(27)
```

```python
# pass statment   take no action
#      used in def 
# application for a bank (  mutual funds)  money in bond market 
#  launch a scheme

#  if amount invetsed is more than 100000 $    give 8 % return
# if amount is less than 100000 then company 
#                has not decided the rate of intrest
```

```python
def bond_rates(amout_invest):
    if amout_invest < 100000:
        pass
    else :
        print("intrest rate is 8 %")
        print(amout_invest * .08 * 1 , " interest given in 1 year")
```

```python
l4 = []
region = [ 'north' , 'south' , 'east' , 'west',]
sales  = [   200 ,    350 ,     50 , 150  ]
profit = [    80 ,    170 ,     10 , 40]
reg_sal  = zip(region , sales, profit)
for i in reg_sal:
    l4.append(i)
```

```python
#  predictive model-->
# model will do a prediction

# any model will predict only one variable at a time 
#  variable which is being preidctied is know as Y  ( dependent )  / Target  variable 

# and all other variables( columns) become X ( independent ) variable 
#  that there can be more than 1 X variable 

```

```python
#  steps involved in model building
# 1 requirment gathering / problem statmemt  u know your Y variable
# 2 data gathering / mining  / collection
# 3 data cleaning / pre processing 
# 4  sampling( divide data into train data and test data )
# 5  Training the model ( build the model)
# 6 testing the model 
# 7 if needed rerun from any previous step to imporve the model



```

```python
# sampling( divide data into train data and test data )
#  q1 why to divide into 2 parts  ( answered)
#  q2 ratio of division    (  more in train and less in test)  
#  q3 how to divide   randomly 
```

```python
def remove_outliers(df,  col,  k):   
    mean = df[col].mean()
    global df1
    sd = df[col].std()    
    final_list = [x for x in df[col] if (x > mean - k * sd)]
    final_list = [x for x in final_list if (x < mean + k * sd)]       
    df1 = df.loc[df[col].isin(final_list)] ; print(df1.shape)     
    print("Number of outliers removed == >"   ,  df.shape[0]  - df1.shape[0])
```

```python
# Class Imbalance  ---> a problem ( issue)

#  is scenario where more represetation of one class
#  less for another

#  model performs good for over represented class
# fails for under represented class
```

```python
#  Class Imbalance is problemm
# solution is 1 over sampling( up sampling)
#             2 under sampling  ( down sampling)           
```

```python
#  over sampling is to increase  the records of class which are under represented 
#  train data 

#  how to increase        ?
#  why only on train data  ?
```

```python
#  under sampling  decreae the records of the class which is over reprsented 
# 
```

```python
#  OS on full data ( before train test split )

rec1 ---> 3 copies 
```

```python
# in os we add duplicte records for the class which is under represented
# and since we are adding duplicates it should be done only on train

#     bcoz testing should not be done on duplicates

# #
# if os is done before train test split ( sampling)  same record may be part of train
# and may be part of test
```

```python
#  under sampling 
#  descreas records class which is over represented 
#  train test split
#  test data 
```

```python
# exam ch1, ch2   ...ch5
#  CH3  all questions are from ch3
#       testing should be done on all the possible scenarios 
```

```python
#  confusion matrix 
```

```python
# so make duplicate entries of the under represented class 
#  add in train data 

#  why only on train data 
```

```python
def remove_outliers(df,  col,  k):   
    mean = df[col].mean()
    global df1
    sd = df[col].std()    
    final_list = [x for x in df[col] if (x > mean - k * sd)]
    final_list = [x for x in final_list if (x < mean + k * sd)]       
    df1 = df.loc[df[col].isin(final_list)] ; print(df1.shape)     
    print("Number of outliers removed == >"   ,  df.shape[0]  - df1.shape[0])
```

```python
#  class imbalance is problem
#  more records for one class and less for another
# in such scenarios model does good for class which is over represented( more records)
# in such scenarios model fails  for class which is under represented(less records)

#  over  sampling
# increase the records for class which is under represneted but only on train data

#  under sampling
# decrease the records for class which is over represneted but only on train data
```

#  checking the stationarity of the data 

```python
#  stationruty is where mean and variance are time invariant ( dont change with time )
```

```python
#  ADF  test   is used to determine if series is stationary or not 
#  augumented dickey fuller test 

#  in adf ho  :--->                    is series is NOT Stationary
#  in adf ha  :--->                    is series is     Stationary
```

```python
from statsmodels.tsa.stattools  import adfuller
```

```python
ap = pd.read_csv(r"C:\Users\pranj\OneDrive\Documents\AirPassengers.csv")
ap['Month'] =  pd.to_datetime(ap['Month'] )
ap = ap.set_index(['Month'])
ap.head(2)
```

```python
adfuller(ap)
#  pvalue is .99 > .05  
# fail to reject the null ( null is still true)
#  conclusion is Series is not stationary
```

```python
# lets check after level 1 diff
```

```python
adfuller(ap.diff().dropna())
```

```python
0.054213290283826474  > .05 
#  fail to reject the null 
```

```python
adfuller(ap.diff().diff().dropna())
#   in adf ho  :--->                    is series is NOT Stationary
#   in adf ha  :--->                    is series is     Stationary

#  after adf testing p values is less than .05  so null is rejected 
#  conclusion is after level 2 diffrincing series becoemes stationary 
```

```python
#  buidling the elbow plot
#  we want ssd values for resptive K values 
```

```python
ssd_list= []
for i in range(1 , 11):
    km = KMeans(n_clusters=i)
    km.fit(mall)
    ssd =km.score(mall)
    ssd_list.append(ssd)    
```

```python
ssd_list = np.absolute(ssd_list)
```

```python
#  plot of elbow plot  K is on X axis and SSD on Y axis 
```

```python
k = list(range(1, 11))
k         
```

```python
plt.figure(figsize=(12, 6))
plt.plot(k  , ssd_list , marker ="*", mec ='red')
plt.xlabel(" Number of customer")
plt.ylabel(" SSD")
plt.title("   elbow plot on customer data" , size = 15);
```

# code for learning rate  decay

```python
model = Sequential()
model.add(Conv2D(filters = 64 , kernel_size =(5,5) , activation ='relu'  ,    padding= 'same', input_shape =( 28 , 28,1)))
model.add(MaxPool2D(pool_size =(2,2)))

model.add(Conv2D(filters = 32 , kernel_size =(5,5) , activation= 'relu'))
model.add(MaxPool2D(pool_size =(2,2)))
model.add(Flatten())

model.add(Dense(128 , activation = 'relu'))
model.add(Dropout(0.4))
model.add(Dense(10 , activation= 'softmax'))
```

```python
#    anither way of doing the decay

#  in the method  learning rate will start to decay when validation loss either remains same of starts to increase
#  user can decicde for how many epochs

from keras.datasets import fashion_mnist
(fas_train_x , fas_train_y ), (fas_test_x, fas_test_y )= fashion_mnist.load_data()  #  loading  the data 

fas_train_y = to_categorical(fas_train_y)
#------------------------  creating the sturcture of CNN  starts 
model = Sequential()
model.add(Conv2D(filters = 64 , kernel_size =(5,5) , activation ='relu'  ,    padding= 'same', input_shape =( 28 , 28,1)))
model.add(MaxPool2D(pool_size =(2,2)))

model.add(Conv2D(filters = 32 , kernel_size =(5,5) , activation= 'relu'))
model.add(MaxPool2D(pool_size =(2,2)))
model.add(Flatten())

model.add(Dense(128 , activation = 'relu'))
model.add(Dropout(0.4))
model.add(Dense(10 , activation= 'softmax'))

#------------------------  creating the sturcture of CNN  ends


#  LR decay code 

from tensorflow.keras.callbacks  import ReduceLROnPlateau

reduce_lr = ReduceLROnPlateau(  monitor='val_loss' , factor=0.5 , patience= 2 ,  min_lr=.000006)
# patience= 2  if for 2 epochs val_loss remains same or increase then reduce the lr 
#  reduce the lr by  factor of .5
#  lowest value of lr is  .000006
model.compile(optimizer='adam', loss = 'categorical_crossentropy' , metrics = ['accuracy'])
model.fit(fas_train_x , fas_train_y , validation_split =.2 , epochs = 15 , callbacks = [reduce_lr] )
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense


(fas_train_x, fas_train_y), (fas_test_x, fas_test_y) = fashion_mnist.load_data()


fas_train_x = np.expand_dims(fas_train_x, -1)   # (60000, 28, 28, 1)
fas_test_x  = np.expand_dims(fas_test_x, -1)

# 3. Resize to 32x32 and convert to 3 channels (RGB)
fas_train_x = tf.image.resize(fas_train_x, [32, 32])
fas_test_x  = tf.image.resize(fas_test_x, [32, 32])
fas_train_x = tf.image.grayscale_to_rgb(fas_train_x)  # (60000, 32, 32, 3)  #  as  mnist fasion was grey scale
fas_test_x  = tf.image.grayscale_to_rgb(fas_test_x)


fas_train_x = fas_train_x / 255.0

fas_train_y = to_categorical(fas_train_y, num_classes=10)
fas_test_y  = to_categorical(fas_test_y, num_classes=10)


base_model = VGG16(include_top=False, weights='imagenet', input_shape=(32, 32, 3))


model = Sequential()
model.add(base_model)
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax')) 


model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])


model.fit(fas_train_x, fas_train_y, validation_split=0.2, epochs=2)
```

```python
#  GLOVE model is pre trained by standord university
```

```python
print('Indexing word vectors.')
embeddings_index = {}
f = open(r"E:\glove.6B.100d.txt",  encoding = 'utf8')  ##  u need to download

for line in f:
    values =  line.split()
    word   =  values[0]
    coefs = np.asarray(values[1:] , dtype= 'float32')
    embeddings_index[word] = coefs
f.close()

print('Found %s word vectors,' %len(embeddings_index))
```

```python
u = embeddings_index['wheat']

norm_u  = np.linalg.norm(u)   ## normanlizing the vector ( take the square root of the sum of square)
similarity = []

for word in embeddings_index.keys():   ## it searches in all the words in the corpus and calculates sosine similiraiy
    v = embeddings_index[word]
    cosine = np.dot(u , v) /norm_u /np.linalg.norm(v)  ## calculating the cosine similarity
    similarity.append((word ,cosine))
    
    
print(len(similarity))    
```

```python
sorted(similarity ,reverse= True , key = lambda y: y[1] )[:100] 
```

```python
df = pd.read_csv(r"C:\Users\pranj\OneDrive\Documents\spam1.csv" ,  encoding='cp1252')
```

```python
df.head(2)
```

```python
# Problem  statment
# predict if email is ham or spam 

#  RNN to build the model
```

```python
df = df.iloc[: , [0,1]] #  get rid of unwanted cols
df.head(2)
```

```python
df.rename(columns ={'v1': 'Y' , 'v2': 'X'},  inplace = True)
df.head(1)
```

```python
df.Y.replace({'ham':0 , 'spam':1 } , inplace = True);
```

```python
from sklearn.model_selection  import train_test_split
df_train , df_test  = train_test_split(df, test_size= .2)
```

```python
df_train_x  = df_train.iloc[: , 1]
df_train_y  = df_train.iloc[: , 0]


df_test_x  = df_test.iloc[: , 1]
df_test_y  = df_test.iloc[: , 0]
```

```python
max_num_words  = 10000  #  total number of words in collection of documents ( corpus)  hyper paramter
seq_len        = 50     #  how many words from each document ( Hyper paramter)
embedding_size = 100    #  hyper paramter for each word i want 100 similar words 
```

```python
from tensorflow.keras.preprocessing.text  import Tokenizer #  used to assign the numbers to the word
from keras.preprocessing.sequence         import  pad_sequences
```

```python
tokenizer = Tokenizer(num_words=max_num_words )
tokenizer.fit_on_texts(df.X)  #  will assign the id to each word based on freq
df_train_x = tokenizer.texts_to_sequences(df_train_x)
df_train_x  = pad_sequences(df_train_x , maxlen = seq_len ) 
```

```python
df_test_x = tokenizer.texts_to_sequences(df_test_x)
df_test_x  = pad_sequences(df_test_x , maxlen = seq_len ) 
```

```python
df_train_y = to_categorical(df_train_y)
```

```python
df_train_y
```

```python
from keras.layers  import  Dense , SimpleRNN , LSTM,Bidirectional

from keras.layers import Dense, SimpleRNN,  Embedding
from keras.models  import Sequential

from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection  import train_test_split
```

```python
model = Sequential()
model.add(Embedding(input_dim= max_num_words , input_length =seq_len , output_dim= embedding_size))
model.add(SimpleRNN(32))

model.add(Dense(2 , activation='softmax' ))
adam = Adam(learning_rate=0.001)
model.compile(optimizer='adam' , loss = 'categorical_crossentropy' , metrics =['accuracy'])
```

```python
model.fit(df_train_x, df_train_y, epochs= 10 , batch_size= 32 , validation_split= .1)
```

```python
pred = model.predict(df_test_x)

pred_cat = pred.argmax(axis =1 )
```

```python
from sklearn.metrics  import confusion_matrix, classification_report
```

```python
confusion_matrix(df_test_y, pred_cat)
```

```python
print(classification_report(df_test_y, pred_cat))
```

```python
#  trip advisior
import warnings
warnings.filterwarnings('ignore')
df = pd.read_csv(r"E:\kajal-\kajal\DATASETS_problem statments\Trip_advisor_review.csv")
df.head(2)
```

```python
df.Rating.replace({1: 0 , 2:0 , 3:1 , 4:1 , 5:1} , inplace = True);
```

```python
df.head()
df.rename(columns ={'Review':'X' , 'Rating': 'Y'} , inplace = True)
df.head(1)
```

```python
from sklearn.model_selection  import train_test_split
df_train , df_test  = train_test_split(df, test_size= .2)
df_train_x  = df_train.iloc[: , 0]
df_train_y  = df_train.iloc[: , 1]


df_test_x  = df_test.iloc[: , 0]
df_test_y  = df_test.iloc[: , 1]
```

```python
max_num_words  = 10000  #  total number of words in collection of documents ( corpus)  hyper paramter
seq_len        = 50     #  how many words from each document ( Hyper paramter)
embedding_size = 100    #  hyper paramter for each word i want 100 similar words 

tokenizer = Tokenizer(num_words=max_num_words )
tokenizer.fit_on_texts(df.X)  #  will assign the id to each word based on freq
df_train_x = tokenizer.texts_to_sequences(df_train_x)
df_train_x  = pad_sequences(df_train_x , maxlen = seq_len ) 
df_test_x = tokenizer.texts_to_sequences(df_test_x)
df_test_x  = pad_sequences(df_test_x , maxlen = seq_len ) 

df_train_y = to_categorical(df_train_y)

model = Sequential()
model.add(Embedding(input_dim= max_num_words , input_length =seq_len , output_dim= embedding_size))
model.add(Bidirectional(LSTM(128)))

model.add(Dense(2 , activation='softmax' ))
adam = Adam(learning_rate=0.001)
model.compile(optimizer='adam' , loss = 'categorical_crossentropy' , metrics =['accuracy'])

model.fit(df_train_x, df_train_y, epochs= 2 , batch_size= 32 , validation_split= .1)
```

```python
pred = model.predict(df_test_x)
```

```python
confusion_matrix(df_test_y , pred.argmax(axis =1))
```

```python
print(classification_report( df_test_y , pred.argmax(axis =1)))
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
#stopwords.words('english')
```

```python
import string
string.punctuation
```

```python
# before building the TDM stopwords and punctuations need to be removed 
```

```python
stopwords.words('english')
```

```python
def text_process(mess):            ### creating a function
    """                                                        ## a docstring
    1. remove the punctuation
    2. remove the stopwords
    3. return the list of clean textwords
    
    """
    nopunc = [char for char in mess if char not in string.punctuation] #  remove puncations
    nopunc = "".join(nopunc)
    
    return [ word for word in nopunc.split() if word not in stopwords.words('english')] # remove stopwords
```

```python
#df.X.apply(text_process)
```

```python
#df.head()
```

```python
from sklearn.feature_extraction.text  import  CountVectorizer
```

```python
var1 = CountVectorizer(analyzer= text_process).fit(df.X)
```

```python
len(var1.vocabulary_)
#  it implies there are totally 9422 words and each word is represented by a number 
```

```python
# using var1 TDM has to be created 
```

```python
tdm = var1.transform(df.X)
```

```python
tdm.shape
```

```python
# using this TDM we can build the model
# THIS tdm acts as our final Data frame 
```

```python
from sklearn.model_selection  import train_test_split
tdm_train , tdm_test , Y_train , Y_test = train_test_split(tdm , df.Y , test_size= .2)
```

```python
from sklearn.linear_model  import LogisticRegression
logreg = LogisticRegression()
logreg.fit(tdm_train  , Y_train)
```

```python
pred = logreg.predict(tdm_test)
```

```python
from sklearn.metrics  import *
```

```python
tab = confusion_matrix(Y_test , pred)
tab
```

```python
tab.diagonal().sum() / tab.sum()
```

```python
from sklearn.tree  import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(tdm_train  , Y_train)
pred = dt.predict(tdm_test)
tab = confusion_matrix(Y_test , pred)
tab
```

```python
from sklearn.ensemble  import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(tdm_train  , Y_train)
pred = rf.predict(tdm_test)
tab = confusion_matrix(Y_test , pred)
tab
```

```python
tab.diagonal().sum() / tab.sum()
```

```python
sample_image =  np.zeros(shape = ( 500 , 500, 3))
def create_circle(event , x, y , flags , param):
    if event ==cv2.EVENT_LBUTTONDOWN:     #  circle will be drawn at a event( event is left mouse click)
        cv2.circle(sample_image , (x,y), 50 , (255, 255, 255) , 0)


cv2.namedWindow(winname = "sample_image")  
cv2.setMouseCallback("sample_image", create_circle)

while True:
    cv2.imshow("sample_image", sample_image)
    if cv2.waitKey(30)  & 0xff == 27: 
        break

cv2.destroyAllWindows()       
```

```python
sample_image =  np.zeros(shape = ( 500 , 500, 3))
def create_circle(event , x, y , flags , param):
    if event ==cv2.EVENT_LBUTTONDOWN:     #  circle will be drawn at a event( event is left mouse click)
        cv2.circle(sample_image , (x,y), 50 , (255, 0, 0) , -1)

    if event ==cv2.EVENT_RBUTTONDOWN:  
        cv2.circle(sample_image , (x,y), 25 , (0, 0, 255) , -1)


cv2.namedWindow(winname = "sample_image")  
cv2.setMouseCallback("sample_image", create_circle)

while True:
    cv2.imshow("sample_image", sample_image)
    if cv2.waitKey(30)  & 0xff == 27: 
        break

cv2.destroyAllWindows()       
   
```

```python
#  breeak 
image = cv2.imread(r"C:\Users\pranj\OneDrive\Desktop\child.JPG" )
```

```python
#plt.imshow(image)
```

```python
face_cascade =  cv2.CascadeClassifier(r"G:\Harcascade\haarcascade_frontalface_default.xml") # pre trained model
#                                                                                          # u need to down load
```

```python
def  face_detection(img):
    face_img = img.copy()
    
    face_rects = face_cascade.detectMultiScale(face_img) # this face_rects is just x and y postion and height & breadth of rectangle
    
    for ( x, y , w, h) in face_rects:
        cv2.rectangle(face_img  ,(x, y) , (x+w , y+h) , (255, 0, 255) , 5)  # drawing the red rectangle on face
    return face_img    
    
```

```python
image_with_rect = face_detection(image) # call the pgm
```

```python
eyes_cascade =  cv2.CascadeClassifier(r"G:\Harcascade\haarcascade_eye.xml")
```

```python
def  eye_detection(img):
    face_img = img.copy()
    
    eye_rects = eyes_cascade.detectMultiScale(face_img) # this face_rects is just x and y postion and height & breadth of rectangle
    
    for ( x, y , w, h) in eye_rects:
        cv2.rectangle(face_img  ,(x, y) , (x+w , y+h) , (255, 0, 0) , 5)  # drawing the red rectangle on face
    return face_img    
```

```python

import cv2 
face_cascade =  cv2.CascadeClassifier(r"G:\Harcascade\haarcascade_frontalface_default.xml")
eyes_cascade =  cv2.CascadeClassifier(r"G:\Harcascade\haarcascade_eye.xml")
cap = cv2.VideoCapture(0) 
while 1:     
    ret, frame = cap.read() 
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY) 
    faces = face_cascade.detectMultiScale(gray, 1.3, 5) 
    for (x,y,w,h) in faces:         
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,255,0),2) 
        roi_gray = gray[y:y+h, x:x+w] 
        roi_color = frame[y:y+h, x:x+w] 
        eyes = eyes_cascade.detectMultiScale(roi_gray) 

        for (ex,ey,ew,eh) in eyes:         
            cv2.rectangle(roi_color,(ex,ey),(ex+ew,ey+eh),(0,127,255),2) 
        cv2.imshow('My Frame work',frame) 
    k = cv2.waitKey(30) & 0xff
    if k == 27:
            break
cap.release() 
cv2.destroyAllWindows()

#  3 loops
# 1st while is till the time camera is open
#  1st for loop is till the time it gets face
#  2nd for is for eyes 
```

```python

def  face_extractor(img):
    faces = face_cascade.detectMultiScale(img, scaleFactor = 1.5 , minNeighbors = 5)
    if faces is ():
        return None
    
    
    for (x ,y , w, h)  in faces:
        cv2.rectangle(img , (x,y) , (x+w , y+h), (0,0, 255) , 2)
        roi =  img[y :y+h , x:x+w]
    
    return roi
```

```python
def voice_to_text(ask = False):   ##  ask = False  it is a default parameter but if we do not give it gives error
    with  sr.Microphone()  as source:
        if ask:
            print(ask)
        print("Say something dear:::")
        audio = r.listen(source)
        try:
            text= r.recognize_google(audio)    # there are many apis like bing , ibm...but i am using the google api
            print("Audio was --->" , text)
        except:
            print("Not able to recognize the voice and convert it")                    
        return(text)
          
```

```python
df =apriori(df_for_mb , min_support=0.025 , use_colnames=True) # .03 means 3 %
#df
```

```python
20 / 300  for  freq product
```

```python
rules =association_rules(df , metric='confidence' ,min_threshold=.5)

#  these rules have to be read / studied and if there is any rule   action might be taken
```

```python
#rules[5:10]
```

```python
#  lift is calculated for product A ---> B 

# if some one buys Prod A lift indicated wether the chance of buying prod B have increased or decreased
# if lift is  greater than 1 then it implies chance of buying prod B have increased 

# if lift is  less than 1 then it implies chance of buying prod B have decreased 

#  formula for lift Lift(A --> B) ===> Conf(A--> B) /Support(B)

# example if lift of (A --> B) is 1.25  this implese chance of buying product B have increase by 25%  after product A was bought

# higher the lift more are the chance of buying prod B after prod A was bought
```

```python
#  formula for leverage

# leverage( A---> B) = Support( A, B) -   ( Support (A) * Support (B))

# leverage is like your LIFT   but advantage of lift is that its range bound  between and -1 and +1

#  if the value is +ve it implies chance of buying B have increase bcoz prod A was bought

#  if the value is -ve it implies chance of buying B have decreased  bcoz prod A was bought
```

```python
#  conviction
# if the rule was created by chance or not 
#  formula 
#  Conviction( A--> B)= (1- Support(B) )  / (1 - Conf(A--> B))

#  if value of conviction is High  it means  consequents(B)  is highly dependent on  antecedents( A) 

# if value is 1 the it means they are independent
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics  import confusion_matrix , classification_report

import warnings
warnings.filterwarnings('ignore')
```

```python
bp = pd.read_csv(r"C:\Users\pranj\OneDrive\Documents\Patient_with_abnormal_bloodpressure.csv")
```

```python
bp.head()

```

```python
bp.isnull().sum()[bp.isnull().sum()>0]
```

```python
bp.Genetic_Pedigree_Coefficient.fillna(bp.Genetic_Pedigree_Coefficient.mean() , inplace = True)
bp.Pregnancy.fillna(0 , inplace = True)
bp.alcohol_consumption_per_day.fillna(bp.alcohol_consumption_per_day.mean() , inplace = True)
```

```python
bp =bp.drop(['Patient_Number'] , axis =1)
```

```python
bp.isnull().sum()[bp.isnull().sum()>0]
```

```python
from sklearn.model_selection  import train_test_split

bp_train , bp_test = train_test_split(bp  , test_size= .2 , random_state= 4729)
bp_train_x  = bp_train.iloc[: , 1::]
bp_train_y  = bp_train.iloc[: , 0]

bp_test_x  = bp_test.iloc[: , 1::]
bp_test_y  = bp_test.iloc[: , 0]
```

```python
#  logistic 
# Dt 
# Rf
# Ada boost
# Gradient
# XGB
#  Cat boost
```

```python
# measuring the similarity on text data
```

```python
#  2 texts messages find similarty 
```

```python
A = "This is an example to show if strings match"
B = "string match or not  this is just an example to show"
```

```python
A =  A.lower()
B =  B.lower()
```

```python
# instrection of  A  and B
```

```python
A =set(A.split())
A
```

```python
B =set(B.split())
```

```python
A
```

```python
B
```

```python
len(A.intersection(B))
```

```python
#  aim was to find similarty between A and B
```

```python
print( len(A) , len(B))
```

```python
len(A.intersection(B))  /  len(B)

# similarty using over lap coef 
```

```python
over lap coef    = len(intersection) /( len of max(a) or (b) )
```

```python
# higher the value more is the similarity
```

```python
def overlap_coef( A, B):
    A =  A.lower()
    B =  B.lower()
    A =set(A.split())
    B =set(B.split())
    global common_set
    common_set = A.intersection(B)
    inter_length = len(A.intersection(B))
    if len(A) > len(B):       
        overlap_coef =  inter_length  / len(A)
    else:
        overlap_coef =  inter_length  / len(B)
    print("similarty score is " ,     overlap_coef )  
```

```python
A = "movie was not good it was     a waste of time"
B = "movie was not bad  it was not a waste of time"
overlap_coef( A, B)
```

```python
def overlap_coef( A, B):
    A =  A.lower()
    B =  B.lower()
    A =set(A.split())
    B =set(B.split())    
    inter_length = len(A.intersection(B))
    union_length = len(A.union(B))
    print(" Jaccard similarty score is " ,     inter_length  /union_length  )  
```

```python
A = "movie was not good it was     a waste of time"
B = "movie was not bad  it was not a waste of time"
overlap_coef( A  , B)
```

```python
#  dice  sorensen   formula is  2* len(A.intersection(B))  /   (  len(A )  + len( B) )
def dicesorensen_coef( A, B):
    A =  A.lower()
    B =  B.lower()
    A =set(A.split())
    B =set(B.split())    
    inter_length = len(A.intersection(B))   
    print(" dice sorensen similarty score is " ,     2 *inter_length  /(len(A) + len(B)  )  )
```

```python
A = "movie was not good it was     a waste of time"
B = "movie was not bad  it was not a waste of time"
dicesorensen_coef( A  , B)
```

```python
#  lets put all code together

# run our (replicate on more than 1 sentence)
#  problem statment 
#  we have a data set with many documents and we want to
# find which document is more similar to doc given by user

text_data = pd.DataFrame()
text_data['Seq_no'] = [1,2,3,4,5,6,7,8]
text_data['Text']   = [ ["Linear Regression is a regression algo"],
                        ["Adaboost is is tree based ensemble  and  classification algo"] ,
                        ["Logistic Regression is a regression and classification algo"], 
                        ["Decision  tree is a tree based classification algo"],
                        ["Random forest is  a tree based ensemble  bagging  algo"] ,
                        ['Ridge is also a regression algo'],
                        ['lasso is also a regression algo'],
                        ['SVM can also be used for regression'] ]


def dicesorensen_coef( A, B):
    global  score_list
    score_list = []
    A =  A.lower()
    A = A.split()
    A = set(A)
    for i in  B:
        B = set(i[0].lower().split())
        dice_soren_score = 2* len(A.intersection(B) ) / ( len(A)  + len(B) )
        score_list.append(dice_soren_score)

A ="Classification algo  is used for categories " #  statment given by user  change as  your need

dicesorensen_coef( A,text_data['Text'] ) #  call to the  function

text_data['Score'] = score_list   #  creating a col name score

text_data  = text_data.sort_values('Score', ascending = False)  # sort the data 
text_data
```

```python
#  lets put all code together

# run our (replicate on more than 1 sentence)
#  problem statment 
#  we have a data set with many documents and we want to
# find which document is more similar to doc given by user

text_data = pd.DataFrame()
text_data['Seq_no'] = [1,2,3,4,5,6,7,8]
text_data['Text']   = [ ["Linear Regression is a regression algo"],
                        ["Adaboost is is tree based ensemble  and  classification algo"] ,
                        ["Logistic Regression is a regression and classification algo"], 
                        ["Decision  tree is a tree based classification algo"],
                        ["Random forest is  a tree based ensemble  bagging  algo"] ,
                        ['Ridge is also a regression algo'],
                        ['lasso is also a regression algo'],
                        ['SVM can also be used for regression'] ]


def jaccard_coef( A, B):
    global  score_list
    score_list = []
    A =  A.lower()
    A = A.split()
    A = set(A)
    for i in  B:
        B = set(i[0].lower().split())
        Jaccard_score =  len(A.intersection(B) ) /  len(A.union(B) )
        score_list.append(Jaccard_score)

A ="Classification algo  is used for categories " #  statment given by user  change as  your need

jaccard_coef( A,text_data['Text'] ) #  call to the  function

text_data['Score'] = score_list   #  creating a col name score

text_data  = text_data.sort_values('Score', ascending = False)  # sort the data 
text_data
```

```python
import pandas as pd
mall = pd.read_csv(r"C:\Users\pranj\OneDrive\Documents\mall_kmeans.csv")
```

```python
mall = mall.drop(['CustomerID'] , axis =1)
```

```python
#mall.isnull().sum()
```

```python
mall['Genre']   = mall['Genre'].replace({'Male':1 , 'Female':0}  );
```

```python
from sklearn.cluster   import KMeans
from sklearn.metrics  import silhouette_score
```

```python
km  = KMeans(n_clusters=2)
km.fit(mall)
pred =km.predict(mall)
silhoutee_score= silhouette_score(mall,pred)
silhoutee_score
```

```python
km  = KMeans(n_clusters=3)
km.fit(mall)
pred =km.predict(mall)
silhoutee_score= silhouette_score(mall,pred)
silhoutee_score
```
